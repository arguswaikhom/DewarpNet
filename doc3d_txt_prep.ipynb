```python
import os

project_root = os.path.abspath(os.path.dirname(__file__)) if '__file__' in globals() else os.getcwd()
doc3d_dir = os.path.join(project_root, "data", "doc3d")

if os.path.exists(doc3d_dir):
    print(f"The directory '{doc3d_dir}' exists.")
else:
    print(f"The directory '{doc3d_dir}' does not exist.")
```

```python
img_folder_path = os.path.join(doc3d_dir, "img")
all_image_paths = []

for root, _, files in os.walk(img_folder_path):
    for file in files:
        full_path = os.path.join(root, file)
        relative_path = os.path.relpath(full_path, img_folder_path)
        all_image_paths.append(relative_path)

total_image_count = len(all_image_paths)
print(f"Collected {total_image_count} relative image paths.")
```

```python
img_folder_path = os.path.join(doc3d_dir, "img")
total_image_count = 0

if os.path.exists(img_folder_path):
    for root, _, files in os.walk(img_folder_path):
        for file in files:
            # You can add checks for image file extensions here if needed
            # For example: if file.lower().endswith(('.png', '.jpg', '.jpeg')):
            total_image_count += 1

    print(f"Total number of image files in '{img_folder_path}': {total_image_count}")
else:
    print(f"The directory '{img_folder_path}' does not exist.")
```

```python
import random

random.shuffle(all_image_paths)

print(f"Successfully shuffled {len(all_image_paths)} image paths.")
```

```python
train_count = int(total_image_count * 0.8)
val_count = total_image_count - train_count

print(f"Total images for training (80%): {train_count}")
print(f"Total images for validation (20%): {val_count}")
```

```python
train_image_paths = all_image_paths[:train_count]
val_image_paths = all_image_paths[train_count:]

print(f"Number of training image paths: {len(train_image_paths)}")
print(f"Number of validation image paths: {len(val_image_paths)}")
```

```python
train_file_path = os.path.join(doc3d_dir, "train.txt")
with open(train_file_path, 'w') as f:
    for path in train_image_paths:
        path_without_extension = os.path.splitext(path)[0]
        f.write(path_without_extension + '\n')

val_file_path = os.path.join(doc3d_dir, "val.txt")
with open(val_file_path, 'w') as f:
    for path in val_image_paths:
        path_without_extension = os.path.splitext(path)[0]
        f.write(path_without_extension + '\n')

print(f"Training image paths written to '{train_file_path}'")
print(f"Validation image paths written to '{val_file_path}'")
```
